### Preparación del entorno
Importamos NumPy y PyTorch, fijamos semillas y elegimos CPU o GPU de forma opcional.


```python
import numpy as np
import torch
import matplotlib.pyplot as plt
from torch import nn
import torch.nn.functional as F

SEED = 2026
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.set_num_threads(1)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
plt.rcParams['figure.figsize'] = (7, 4)
print('PyTorch:', torch.__version__, '| dispositivo:', device)
```

### Gradiente explícito NumPy
Iteramos sobre una cuadrática y almacenamos la trayectoria.


```python
parameter_value = 4.0
learning_rate = 0.05
loss_value = (parameter_value - 1.0) ** 2
gradient_value = 2.0 * (parameter_value - 1.0)
next_parameter = parameter_value - learning_rate * gradient_value
print('parámetro inicial:', parameter_value, '| pérdida:', loss_value, '| gradiente:', gradient_value)
print('parámetro tras un paso:', next_parameter)
```

**Resultado e interpretación.** La implementación completa de la actualización aparece en la celda siguiente; evita abstraer el gradiente antes de inspeccionar su forma.


### Trayectorias de optimización
Calculamos derivadas analíticas con NumPy y automáticas con PyTorch para dos tasas.


```python
rates = [0.05, 0.9]
fig, axes = plt.subplots(1, 2, figsize=(10, 3.5))
for ax, rate in zip(axes, rates):
    parameter = 4.0; path = []; losses = []
    for step in range(35):
        path.append(parameter); losses.append((parameter - 1.0) ** 2)
        gradient = 2.0 * (parameter - 1.0)
        parameter -= rate * gradient
    ax.plot(losses); ax.set_title(f'tasa = {rate}'); ax.set_xlabel('Paso'); ax.set_ylabel('Pérdida')
plt.tight_layout(); plt.show()
parameter = torch.tensor(4.0, requires_grad=True)
loss = (parameter - 1.0) ** 2
loss.backward()
print('gradiente autograd en 4:', parameter.grad.item())
```

**Resultado e interpretación.** La trayectoria muestra el efecto de la tasa en esta función concreta. La tasa útil depende de la escala y geometría del problema.
