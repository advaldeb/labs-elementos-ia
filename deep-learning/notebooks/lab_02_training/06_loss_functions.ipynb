## Teoría breve
Regresión: $L_{MSE}=\frac{1}{N}\sum_i(y_i-\hat y_i)^2$. Clasificación: $L_{CE}=-\sum_c y_c\log p_c$. `CrossEntropyLoss` recibe logits, no probabilidades.

### Preparación del entorno
Importamos NumPy y PyTorch, fijamos semillas y elegimos CPU o GPU de forma opcional.


```python
import numpy as np
import torch
import matplotlib.pyplot as plt
from torch import nn
import torch.nn.functional as F

SEED = 2026
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.set_num_threads(1)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
plt.rcParams['figure.figsize'] = (7, 4)
print('PyTorch:', torch.__version__, '| dispositivo:', device)
```

**Resultado e interpretación.** La celda confirma el entorno utilizado. El resto del laboratorio está diseñado para CPU; una GPU disponible se utiliza sin ser requisito.


### Pérdidas de regresión
Calculamos MAE y MSE con y sin una observación extrema.


```python
y_true = torch.tensor([1., 2., 3., 4.])
pred = torch.tensor([1.5, 1.5, 2.5, 5.0])
mae = torch.mean(torch.abs(y_true - pred))
mse = torch.mean((y_true - pred) ** 2)
y_extreme = y_true.clone(); y_extreme[-1] = 20
print('MAE, MSE:', mae.item(), mse.item())
print('con objetivo extremo, MAE/MSE:', torch.abs(y_extreme-pred).mean().item(), ((y_extreme-pred)**2).mean().item())
```

**Resultado e interpretación.** MSE aumenta cuadráticamente con errores grandes; MAE crece linealmente. La elección implica una noción distinta de error costoso.


### Entropía cruzada
Calculamos la pérdida multiclase directamente desde logits.


```python
logits = torch.tensor([[2.0, 0.5, -1.0], [0.1, 0.2, 1.5]])
targets = torch.tensor([0, 2])
ce = nn.CrossEntropyLoss()
loss_from_logits = ce(logits, targets)
probabilities = torch.softmax(logits, dim=1)
print('probabilidades:', probabilities.numpy().round(3))
print('CrossEntropyLoss:', loss_from_logits.item())
print('predicciones:', logits.argmax(dim=1).tolist(), '| etiquetas:', targets.tolist())
```

**Resultado e interpretación.** `CrossEntropyLoss` combina softmax y log-loss de manera numéricamente estable. La métrica de acierto y la pérdida responden preguntas diferentes.
