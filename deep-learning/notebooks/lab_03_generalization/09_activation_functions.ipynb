**Pregunta guía:** ¿Cómo afectan las activaciones a la no linealidad y al flujo de gradientes?

## Objetivos
Al finalizar podrás comparar sigmoid, tanh, ReLU y GELU; visualizar curvas y derivadas; reconocer saturación.

## Prerrequisitos y conexión

## Teoría breve
Sigmoid y tanh se saturan para magnitudes grandes; ReLU es $\max(0,z)$; GELU suaviza el umbral. La derivada determina cuánto gradiente atraviesa una unidad.

## Problema y datos
Evaluamos cuatro activaciones y sus gradientes en una misma rejilla. Todos los datos se generan en el notebook: no se requieren descargas ni archivos locales.

## Experimento conceptual

### Preparación del entorno
Importamos NumPy y PyTorch, fijamos semillas y elegimos CPU o GPU de forma opcional.


```python
import numpy as np
import torch
import matplotlib.pyplot as plt
from torch import nn
import torch.nn.functional as F

SEED = 2026
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.set_num_threads(1)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
plt.rcParams['figure.figsize'] = (7, 4)
print('PyTorch:', torch.__version__, '| dispositivo:', device)
```

**Resultado e interpretación.** La celda confirma el entorno utilizado. El resto del laboratorio está diseñado para CPU; una GPU disponible se utiliza sin ser requisito.


### Curvas y gradientes
Usamos autograd para observar activación y derivada sin aproximaciones manuales.


```python
values = torch.linspace(-6, 6, 300, requires_grad=True)
activations = {'sigmoid': torch.sigmoid(values), 'tanh': torch.tanh(values),
               'ReLU': F.relu(values), 'GELU': F.gelu(values)}
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
for name, output in activations.items():
    axes[0].plot(values.detach(), output.detach(), label=name)
    gradient = torch.autograd.grad(output.sum(), values, retain_graph=True)[0]
    axes[1].plot(values.detach(), gradient.detach(), label=name)
axes[0].set(title='Activación', xlabel='z'); axes[1].set(title='Derivada', xlabel='z')
for ax in axes: ax.legend(); ax.grid(alpha=.2)
plt.tight_layout(); plt.show()
```

**Resultado e interpretación.** La gráfica permite localizar saturación y cambios de pendiente. La función apropiada depende de arquitectura, inicialización y objetivo, no solo de la forma aislada.
